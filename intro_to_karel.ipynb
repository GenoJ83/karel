# Introduction to Stanford Karel

Welcome to Karel the Robot! Karel is a great way to learn programming. Karel understands a few basic commands and lives in a grid world.

## 1. Setup
First, we need to import Karel.

```python
from stanfordkarel import *
print('Karel library imported successfully!')
```

## 2. Basic Commands
Karel knows exactly four basic commands:
- `move()`: Karel moves one block forward.
- `turn_left()`: Karel turns 90 degrees to the left.
- `pick_beeper()`: Karel picks up a beeper from the ground.
- `put_beeper()`: Karel places a beeper on the ground.

Let's write a simple program!

```python
def main():
    move()
    put_beeper()
    move()
    turn_left()
    move()

if __name__ == '__main__':
    run_karel_program()
```

## 3. The While Loop (Move to Wall)
Instead of counting steps, we can use a `while` loop to make Karel walk until it hits a wall! We use the `front_is_clear()` sensor for this.

```python
def main():
    # As long as there is no wall directly in front of Karel...
    while front_is_clear():
        # Take one step forward
        move()
        
    # When the loop finishes, the program ends.

if __name__ == '__main__':
    run_karel_program()
```

## 4. The Fencepost Bug (Fill the Row)
Let's put a line of beepers across the floor. Watch out for a common trap: the loop stops exactly when Karel hits the wall, so it forgets to put a beeper on the very last square!

```python
def main():
    while front_is_clear():
        put_beeper()
        move()
        
    # We must manually place one last beeper at the end!
    put_beeper()

if __name__ == '__main__':
    run_karel_program()
```

## 5. The If Statement (Beeper Sweeper)
Karel can use an `if` statement to check `beepers_present()` and clean up scattered beepers.

```python
def main():
    while front_is_clear():
        # Check if there is a beeper on this square
        if beepers_present():
            pick_beeper()
        
        # Keep moving forward regardless
        move()
        
    # Check the very last square just in case!
    if beepers_present():
        pick_beeper()

if __name__ == '__main__':
    run_karel_program()
```

## 6. Teaching Karel New Words
Karel doesn't know how to `turn_right()`. But we can teach it by combining three `turn_left()` commands in a row using a function!

```python
def turn_right():
    turn_left()
    turn_left()
    turn_left()

def main():
    move()
    turn_right()
    move()

if __name__ == '__main__':
    run_karel_program()
```
