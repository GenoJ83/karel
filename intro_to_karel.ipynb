# Introduction to Stanford Karel

Welcome to Karel the Robot! Karel is a great way to learn programming. Karel understands a few basic commands and lives in a grid world.

## 1. Setup
First, we need to import Karel.

```python
from stanfordkarel import *
print('Karel library imported successfully!')
```

## 2. Basic Commands
Karel knows exactly four basic commands:
- `move()`: Karel moves one block forward.
- `turn_left()`: Karel turns 90 degrees to the left.
- `pick_beeper()`: Karel picks up a beeper from the ground.
- `put_beeper()`: Karel places a beeper on the ground.

Let's write a simple program!

```python
def main():
    move()
    put_beeper()
    move()
    turn_left()
    move()

# Run the program in a standard 10x10 world
if __name__ == '__main__':
    run_karel_program()
```

## 3. Teaching Karel New Words
Karel doesn't know how to `turn_right()`. But we can teach it by combining three `turn_left()` commands in a row using a function!

```python
def turn_right():
    turn_left()
    turn_left()
    turn_left()

def main():
    move()
    turn_right()
    move()

if __name__ == '__main__':
    run_karel_program()
```

## 4. Conditions
Karel can ask questions about its world to decide what to do next. For example, `front_is_clear()` checks if there is a wall in front of Karel.

```python
def main():
    # Move forward safely until a wall is hit
    while front_is_clear():
        move()
    print('I hit a wall!')

if __name__ == '__main__':
    run_karel_program()
```
